```python
import pandas as pd

df = pd.read_parquet(
    "../data/processed/medicines_phase3.parquet"
)

print("Shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print(col)

print("\nData types:")
print(df.dtypes)
```

```python
identity_columns = [
    col for col in df.columns
    if any(
        word in col.lower()
        for word in [
            "canonical",
            "identity",
            "composition",
            "form",
            "modifier",
            "hash"
        ]
    )
]

print("Potential identity columns:")
print(identity_columns)

display(df[identity_columns].head(10))
```

```python

print("ALL DATASET COLUMNS:")
print(df.columns.tolist())

print("\nPOSSIBLE IDENTITY COLUMNS:")
for col in df.columns:
    if any(
        word in col.lower()
        for word in [
            "key", "canonical", "hash",
            "signature", "group", "match"
        ]
    ):
        print(col)

inspect_cols = [
    col for col in [
        "medicine_id",
        "name",
        "price",
        "is_discontinued",
        "dosage_form_normalized",
        "formulation_modifier",
        "formulation_detail",
        "formulation_status",
        "pack_quantity",
        "pack_unit",
        "price_per_unit"
    ]
    if col in df.columns
]

display(df[inspect_cols].head(15))

print("\nNORMALIZED DOSAGE FORM COUNTS:")
print(df["dosage_form_normalized"].value_counts(dropna=False).head(30))

print("\nPACK UNIT COUNTS:")
print(df["pack_unit"].value_counts(dropna=False).head(30))


print("\nFORMULATION STATUS:")
print(df["formulation_status"].value_counts(dropna=False))

print("\nFORMULATION DETAIL:")
print(df["formulation_detail"].value_counts(dropna=False))
```

```python


print("DATA TYPE:")
print(df["parsed_ingredients"].dtype)

print("\nFIRST 10 PARSED INGREDIENT VALUES:")

for i, value in enumerate(df["parsed_ingredients"].head(10)):
    print(f"\nRecord {i + 1}:")
    print(repr(value))

print("\nMISSING PARSED INGREDIENTS:")
print(df["parsed_ingredients"].isna().sum())

print("\nINGREDIENT COUNTS:")
print(df["ingredient_count"].value_counts(dropna=False).sort_index())

print("\nFORMULATION MODIFIER VALUES:")
print(df["formulation_modifier"].value_counts(dropna=False).head(30))
```

```python
from pathlib import Path

source_files = [
    Path("../src/cleaning.py"),
    Path("../src/pricing.py")
]

for file in source_files:
    print(f"\n{'=' * 60}")
    print(f"FILE: {file}")
    print("=" * 60)

    if file.exists():
        content = file.read_text(encoding="utf-8")

        for line_no, line in enumerate(content.splitlines(), start=1):
            if any(
                word in line.lower()
                for word in [
                    "canonical",
                    "signature",
                    "sha256",
                    "hash",
                    "identity"
                ]
            ):
                print(f"{line_no}: {line}")
```

```python

strength_details = []

for ingredients in df["parsed_ingredients"].head(1000):
    for item in ingredients:
        strength_details.append({
            "ingredient": item["ingredient"],
            "strength_value": item["strength_value"],
            "strength_unit": item["strength_unit"],
            "denominator_value": item["denominator_value"],
            "denominator_unit": item["denominator_unit"],
            "strength_basis": item["strength_basis"],
            "strength_qualifier": item["strength_qualifier"],
            "parse_status": item["parse_status"]
        })

strength_df = pd.DataFrame(strength_details)

display(strength_df.head(20))

print("\nSTRENGTH UNITS:")
print(strength_df["strength_unit"].value_counts(dropna=False))

print("\nSTRENGTH BASES:")
print(strength_df["strength_basis"].value_counts(dropna=False))

print("\nSTRENGTH QUALIFIERS:")
print(strength_df["strength_qualifier"].value_counts(dropna=False))

print("\nPARSE STATUS:")
print(strength_df["parse_status"].value_counts(dropna=False))
```
