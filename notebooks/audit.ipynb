```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("../data/raw/Raw_A_Z_medicines_dataset_of_India.csv")
print("Dataset shpae:", df.shape)
display(df.head(10))

print("\n Column Nmes: ")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
audit_df=df.copy()

column_mapping = {
    "id": "medicine_id",
    "name": "name",
    "price(₹)": "price",
    "Is_discontinued": "is_discontinued",
    "manufacturer_name": "manufacturer",
    "type": "dosage_form",
    "pack_size_label": "pack_size",
    "short_composition1": "composition1",
    "short_composition2": "composition2"
}

audit_df.rename(columns=column_mapping, inplace=True)

print("Updated Columns:")
print(audit_df.columns.tolist())

display(audit_df.head())
```

```python
missing_count=audit_df.isna().sum()

missing_percentage=(missing_count/len(audit_df))*100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,"Missing_Percentage": missing_percentage.round(2)})

missing_summary = missing_summary.sort_values(
    by="Missing_Percentage",ascending=False)

display(missing_summary)
```

```python
exact_duplicates = audit_df.duplicated().sum()

logical_duplicates = audit_df.duplicated(
    subset=["name", "manufacturer"]).sum()

print("Exact Duplicate Rows:", exact_duplicates)

print(
    "Duplicate Name + Manufacturer Rows:",logical_duplicates)

duplicate_examples = audit_df[
    audit_df.duplicated(subset=["name", "manufacturer"],keep=False)].sort_values(by=["name", "manufacturer"])

display(duplicate_examples.head(20))

```

```python
audit_df["price"] = pd.to_numeric(audit_df["price"],errors="coerce")

print("PRICE STATISTICS")
display(audit_df["price"].describe())

zero_prices = (audit_df["price"] == 0).sum()

negative_prices = (audit_df["price"] < 0).sum()

missing_prices = audit_df["price"].isna().sum()

print("Zero Prices:", zero_prices)
print("Negative Prices:", negative_prices)
print("Missing Prices:", missing_prices)

print("\n20 HIGHEST MEDICINE PRICES")

display(audit_df.nlargest(20, "price")[ ["name", "price", "manufacturer", "pack_size"]])
```

```python
original_unique = audit_df["name"].nunique()

normalized_names = (audit_df["name"].astype("string").str.lower().str.strip())

normalized_unique = normalized_names.nunique()

print("Original Unique Names:", original_unique)

print("Normalized Unique Names:", normalized_unique)

print("Difference:", original_unique - normalized_unique)
```

```python
display(audit_df["is_discontinued"].value_counts(dropna=False))

discontinued_count = (audit_df["is_discontinued"] == True).sum()

active_count = (audit_df["is_discontinued"] == False).sum()

total_count = len(audit_df)

print("Total Medicines:", total_count)
print("Discontinued Medicines:", discontinued_count)
print("Active Medicines:", active_count)

print("Discontinued Percentage:",round((discontinued_count / total_count) * 100, 2),"%")
```

```python
composition_sample = audit_df[ ["name", "composition1", "composition2"]
].sample( n=50,random_state=42)

composition_sample = composition_sample.fillna("")

display(composition_sample)
```

```python
comp1 = audit_df["composition1"].fillna("").astype(str).str.strip()
comp2 = audit_df["composition2"].fillna("").astype(str).str.strip()

print("Empty Composition 1:", (comp1 == "").sum())
print("Empty Composition 2:", (comp2 == "").sum())

combined_composition = (comp1 + " + " + comp2).str.strip(" +")

ingredient_counts = combined_composition.apply(lambda x: len([item.strip()
        for item in x.split("+")
        if item.strip()]))

single_ingredient = (ingredient_counts == 1).sum()
two_ingredients = (ingredient_counts == 2).sum()
three_or_more = (ingredient_counts >= 3).sum()
no_ingredient = (ingredient_counts == 0).sum()

print("\nEstimated Ingredient Distribution")

print("No Ingredients:", no_ingredient)
print("Single Ingredient:", single_ingredient)
print("Two Ingredients:", two_ingredients)
print("Three or More Ingredients:", three_or_more)

print("Percentage with at least one ingredient:",round(((total_count - no_ingredient) / total_count) * 100, 2),"%")
```

```python
print("DOSAGE CATEGORY DISTRIBUTION")

display(audit_df["dosage_form"].value_counts(dropna=False).head(20))

print("\nPACK SIZE EXAMPLES")

display(audit_df["pack_size"].value_counts(dropna=False).head(30))
```

```python

from pathlib import Path


Path("../docs").mkdir(parents=True, exist_ok=True)


active_percentage = round(active_count / total_count * 100, 2)
discontinued_percentage = round(discontinued_count / total_count * 100, 2)


report = f"""
# Medicine Dataset – Data Quality Audit

## 1. Dataset Overview

- Total records: {len(audit_df):,}
- Total columns: {len(audit_df.columns)}
- Dataset source: Raw_A_Z_medicines_dataset_of_India.csv

## 2. Missing Values

- Missing composition1: {audit_df["composition1"].isna().sum():,}
- Missing composition2: {audit_df["composition2"].isna().sum():,}
- Missing composition2 percentage: {audit_df["composition2"].isna().mean() * 100:.2f}%
- Missing values in other columns: 0

**Decision:** composition2 is optional. Records with one ingredient
will be retained.

## 3. Duplicate Records

- Exact duplicate rows: {exact_duplicates:,}
- Duplicate name + manufacturer combinations: {logical_duplicates:,}

**Decision:** Do not remove records based only on name and manufacturer.
Different pack sizes, strengths, or prices may represent valid products.

## 4. Price Quality

- Mean price: ₹{audit_df["price"].mean():,.2f}
- Median price: ₹{audit_df["price"].median():,.2f}
- Minimum price: ₹{audit_df["price"].min():,.2f}
- Maximum price: ₹{audit_df["price"].max():,.2f}
- Zero-price records: {zero_prices:,}
- Negative-price records: {negative_prices:,}
- Missing-price records: {missing_prices:,}

**Decision:** Zero and invalid prices will be flagged and excluded from
price recommendations. High prices will be investigated rather than
automatically deleted.

## 5. Medicine Status

- Active medicines: {active_count:,} ({active_percentage}%)
- Discontinued medicines: {discontinued_count:,} ({discontinued_percentage}%)

**Decision:** Discontinued records will be retained for traceability
but excluded from recommendations.

## 6. Composition Analysis

- Records with one listed ingredient: {single_ingredient:,}
- Records with two listed ingredients: {two_ingredients:,}
- Records with three or more listed ingredients: {three_or_more:,}
- Records with no listed ingredients: {no_ingredient:,}

**Limitation:** The dataset provides two separate composition columns.
The absence of a third listed ingredient does not establish that every
medicine has at most two ingredients in reality.

## 7. Name Normalization

- Unique names before normalization: {original_unique:,}
- Unique names after lowercase and whitespace normalization: {normalized_unique:,}
- Difference: {original_unique - normalized_unique:,}

**Decision:** Names will be standardized for searching, while original
names will be preserved for display.

## 8. Dosage Form and Pack Size

The dosage_form column contains only the value 'allopathy'.
It does not identify the actual pharmaceutical form.

The pack_size column contains descriptions such as strips, bottles,
vials, and tubes, with units including tablets, ml, and gm.

**Decision:** Actual dosage form and release type will be inferred
from medicine names and pack descriptions during the cleaning phase.

## 9. Key Data Quality Issues and Planned Actions

| Issue | Planned action |
|---|---|
| Missing composition2 | Retain as a valid optional field |
| Logical duplicate combinations | Preserve different product variants |
| Zero prices | Flag and exclude from recommendations |
| Very high prices | Investigate outliers; do not delete automatically |
| Inconsistent ingredient names | Normalize using a reviewed synonym dictionary |
| Inconsistent spacing | Standardize text |
| Multiple strength formats | Parse and normalize units carefully |
| Generic dosage category | Infer actual dosage form from available text |
| Discontinued medicines | Exclude from recommendation results |

## 10. Conclusion

The dataset contains a substantial number of medicine records and
provides useful information about product names, prices, manufacturers,
compositions, and packaging.

The main challenge is standardizing composition strengths, ingredient
names, dosage forms, and pack sizes before reliable price comparisons
can be performed.
"""


report_path = Path("../docs/data_quality.md")
report_path.write_text(report, encoding="utf-8")

print("Data quality report created:", report_path)
```
